## 1. Import libraries
NumPy is used for numerical calculations. Pandas is used to read, organise, filter and analyse table data.


```python
import numpy as np
import pandas as pd
```

## 2. Load your dataset
A DataFrame is a table with rows and columns. Keep the CSV in the same folder as this notebook, or update the path.


```python
df = pd.read_csv("Student_14_dataset.csv")
print("First five records:")
display(df.head())
```

## 3. Explore the data
Use `shape`, `columns`, `info()`, `isnull().sum()` and `duplicated().sum()` to inspect the dataset.


```python
# TODO: Display the shape and column names
# TODO: Inspect data types and missing values
# TODO: Count duplicate rows

print("Dataset shape:",df.shape)
print("Column names:", list(df.columns))
df.info()
print("Missing values:")
print(df.isnull().sum())
print("Duplicate rows:",
      df.duplicated().sum())
```

## 4. NumPy practice
Convert one or more numeric columns to a NumPy array. Try `np.mean()`, `np.max()` and `np.min()` where appropriate.


```python
# TODO: Select numeric columns relevant to your project
# TODO: Convert them to a NumPy array using .to_numpy()
# TODO: Calculate average, maximum and minimum

# Select numeric columns
study = df["Study_Hours"].to_numpy()
practice = df["Practice_Hours"].to_numpy()
sleep = df["Sleep_Hours"].to_numpy()
score = df["Test_Score"].to_numpy()

# Calculate average, maximum and minimum

print("Study Hours:")
print("Average:", np.mean(study))
print("Maximum:", np.max(study))
print("Minimum:", np.min(study))

print("\nPractice Hours:")
print("Average:", np.mean(practice))
print("Maximum:", np.max(practice))
print("Minimum:", np.min(practice))

print("\nSleep Hours:")
print("Average:", np.mean(sleep))
print("Maximum:", np.max(sleep))
print("Minimum:", np.min(sleep))

print("\nTest Scores:")
print("Average:", np.mean(score))
print("Maximum:", np.max(score))
print("Minimum:", np.min(score))
```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
# TODO: Create calculated column(s)
# TODO: Filter records using a condition
# TODO: Sort records and display the top/bottom results


# Create calculated columns
df["Total_Hours"] = df["Study_Hours"] + df["Practice_Hours"]
df["Average_Hours"] = df["Total_Hours"] / 2

# Filter students scoring above 80
print("Students scoring above 80:")
display(df[df["Test_Score"] > 80])

# Sort students according to test scores
print("Top 5 students:")
display(df.sort_values(by="Test_Score", ascending=False).head(5))

print("Bottom 5 students:")
display(df.sort_values(by="Test_Score").head(5))
```

## 6. Your individual project challenge
**Challenge:** Calculate average study time and test score; compare study hours with scores; identify top scores.

Write your solution below and explain what the result means.


```python
# TODO: Solve your assigned challenge


# Calculate average study hours and test score
avg_study = df["Study_Hours"].mean()
avg_score = df["Test_Score"].mean()

print("Average Study Hours:", avg_study)
print("Average Test Score:", avg_score)

# Compare study hours with test scores
print("\nStudy Hours and Test Scores:")
display(df[["Name", "Study_Hours", "Test_Score"]])

# Find students with highest scores
print("\nTop 5 Students:")
top_students = df.sort_values(by="Test_Score", ascending=False)
display(top_students[["Name", "Study_Hours", "Test_Score"]].head(5))
```
